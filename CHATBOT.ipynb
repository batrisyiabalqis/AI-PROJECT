```python

while True:
    user = input("Ask me something: ").lower()

    if "name" in user or "nama" in user:
        print("My name is Batrisyia Balqis.")

    elif "age" in user or "umur" in user:
        print("I am 20 years old.")

    elif "study" in user or "belajar" in user:
        print("I am studying Diploma in Software Engineering at GMI.")

    elif "hobby" in user or "hobi" in user:
        print("My hobbies are reading, drawing, and learning new things.")

    elif "location" in user or "tinggal" in user:
        print("I live in Sepang, Selangor.")

    elif "bye" in user or "exit" in user:
        print("Goodbye! Have a nice day!")
        break

    else:
        print("Sorry, I don't understand.")

```
